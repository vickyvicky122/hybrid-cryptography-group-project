```python
import random

ascii_lowercase = 'abcdefghijklmnopqrstuvwxyz'
ascii_uppercase = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'


# QUESTION 1

# First functions to encrypt and decrypt caesar ciphers
# Note that the key for a caeser cypher is just the value by which we shift each letter in the alphabet.
def caesar_cipher(plaintext, key):
    cipher_text = ''
    # iterate through each letter in the plaintext
    for i in plaintext:
        # shift by the value of the key.
        if i in ascii_lowercase:
            index = ascii_lowercase.index(i)
            cipher_text += ascii_lowercase[(index + key) % 26]
        else:
            index = ascii_uppercase.index(i)
            cipher_text += ascii_uppercase[(index + key) % 26]
    return cipher_text


# in order to decrypt a message run the encryption backwards
def decrypt_caesar_cipher(ciphertext, key):
    return caesar_cipher(ciphertext, -1*key)


# Next functions to encrypt and deccrypt vigenere ciphers

def vigenere_cipher(plaintext, key):
    #Note that for the vigenere cipher, the key is a word.
    length = len(key)
    cipher_text = ''
    # iterate through each letter in the plaintext
    for i in range(len(plaintext)):
        # work out the corresponding ceasar key to the letter
        number = ascii_lowercase.index((key[i % length].lower()))
        # compute the new letter
        cipher_text += caesar_cipher(plaintext[i], number)
    return cipher_text

def decrypt_vigenere_cipher(ciphertext, key):
    length = len(key)
    cipher_text = ''
    # iterate through each letter in the ciphertext
    for i in range(len(ciphertext)):
        # work out the corresponding ceasar key to the letter
        number = ascii_lowercase.index((key[i % length].lower()))
        # decrypt the letter
        cipher_text += decrypt_caesar_cipher(ciphertext[i], number)
    return cipher_text


# now a function to get plaintext from text files

def get_plaintext(txt_file_name):
    # open the file
    with open(txt_file_name, "r") as file:
        text = file.read()
    plaintext = ''
    # iterate through and only add letters to the plaintext
    for i in text:
        if i in ascii_lowercase or i in ascii_uppercase:
            plaintext += i
    return plaintext


# Functions to encrypt and decrypt to and from files.
# Each one reads its input from one text file and writes its result to another.

def write_text(txt_file_name, text):
    with open(txt_file_name, "w") as file:
        file.write(text)

def caesar_encrypt_file(input_file_name, output_file_name, key):
    # only the alphabetic content of the input file is encrypted
    plaintext = get_plaintext(input_file_name)
    write_text(output_file_name, caesar_cipher(plaintext, key))

def caesar_decrypt_file(input_file_name, output_file_name, key):
    ciphertext = get_plaintext(input_file_name)
    write_text(output_file_name, decrypt_caesar_cipher(ciphertext, key))

def vigenere_encrypt_file(input_file_name, output_file_name, key):
    plaintext = get_plaintext(input_file_name)
    write_text(output_file_name, vigenere_cipher(plaintext, key))

def vigenere_decrypt_file(input_file_name, output_file_name, key):
    ciphertext = get_plaintext(input_file_name)
    write_text(output_file_name, decrypt_vigenere_cipher(ciphertext, key))


# Finally functions to test the encryption and decryption of the ciphers with random keys.
# Each test encrypts the message file to a new file, decrypts that file to another new file,
# and checks that the decrypted file matches the alphabetic content of the original.

def test_caesar(txt_file_name):
    # first get the plaintext and generate a random key
    plaintext = get_plaintext(txt_file_name)
    key = random.randint(0, 25)
    # then encrypt and decrypt the message, going through files
    encrypted_file_name = 'caesar_encrypted_' + txt_file_name
    decrypted_file_name = 'caesar_decrypted_' + txt_file_name
    caesar_encrypt_file(txt_file_name, encrypted_file_name, key)
    caesar_decrypt_file(encrypted_file_name, decrypted_file_name, key)
    # finally check the decrypted file matches the original input
    if get_plaintext(decrypted_file_name) == plaintext:
        return 'Test passed: ' + txt_file_name + ' -> ' + encrypted_file_name + ' -> ' + decrypted_file_name
    return 'Test failed, key = ' + str(key)

def test_vigenere(txt_file_name):
    # first get the plaintext and generate a random key
    plaintext = get_plaintext(txt_file_name)
    key_len = random.randint(1, len(plaintext))
    key = ''
    for i in range(key_len):
        key += ascii_lowercase[random.randint(0, 25)]
    # then encrypt and decrypt the message, going through files
    encrypted_file_name = 'vigenere_encrypted_' + txt_file_name
    decrypted_file_name = 'vigenere_decrypted_' + txt_file_name
    vigenere_encrypt_file(txt_file_name, encrypted_file_name, key)
    vigenere_decrypt_file(encrypted_file_name, decrypted_file_name, key)
    # finally check the decrypted file matches the original input
    if get_plaintext(decrypted_file_name) == plaintext:
        return 'Test passed: ' + txt_file_name + ' -> ' + encrypted_file_name + ' -> ' + decrypted_file_name
    return 'Test failed, key = ' + key

print(test_caesar('message_heart_of_darkness.txt'))
print(test_vigenere('message_emma.txt'))
```
