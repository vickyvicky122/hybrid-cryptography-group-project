```python
import random

ascii_lowercase = 'abcdefghijklmnopqrstuvwxyz'
ascii_uppercase = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'


# QUESTION 1

# First functions to encrypt and decrypt caesar ciphers
# Note that the key for a caeser cypher is just the value by which we shift each letter in the alphabet.
def caesar_cipher(plaintext, key):
    cipher_text = ''
    # iterate through each letter in the plaintext
    for i in plaintext:
        # shift by the value of the key.
        if i in ascii_lowercase:
            index = ascii_lowercase.index(i)
            cipher_text += ascii_lowercase[(index + key) % 26]
        else:
            index = ascii_uppercase.index(i)
            cipher_text += ascii_uppercase[(index + key) % 26]
    return cipher_text


# in order to decrypt a message run the encryption backwards
def decrypt_caesar_cipher(ciphertext, key):
    return caesar_cipher(ciphertext, -1*key)


# Next functions to encrypt and deccrypt vigenere ciphers

def vigenere_cipher(plaintext, key):
    #Note that for the vigenere cipher, the key is a word.
    length = len(key)
    cipher_text = ''
    # iterate through each letter in the plaintext
    for i in range(len(plaintext)):
        # work out the corresponding ceasar key to the letter
        number = ascii_lowercase.index((key[i % length].lower()))
        # compute the new letter
        cipher_text += caesar_cipher(plaintext[i], number)
    return cipher_text

def decrypt_vigenere_cipher(ciphertext, key):
    length = len(key)
    cipher_text = ''
    # iterate through each letter in the ciphertext
    for i in range(len(ciphertext)):
        # work out the corresponding ceasar key to the letter
        number = ascii_lowercase.index((key[i % length].lower()))
        # decrypt the letter
        cipher_text += decrypt_caesar_cipher(ciphertext[i], number)
    return cipher_text


# now a function to get plaintext from text files

def get_plaintext(txt_file_name):
    # open the file
    with open(txt_file_name, "r") as file:
        text = file.read()
    plaintext = ''
    # iterate through and only add letters to the plaintext
    for i in text:
        if i in ascii_lowercase or i in ascii_uppercase:
            plaintext += i
    return plaintext


# Finally functions to test the encryption and decryption of the ciphers with random keys

def test_caesar(txt_file_name):
    # first get the plaintext and generate a random key
    plaintext = get_plaintext(txt_file_name)
    key = random.randint(0, 25)
    # then encrypt and deccrypt the message
    ciphertext_test =  caesar_cipher(plaintext, key)
    plaintext_test = decrypt_caesar_cipher(ciphertext_test, key)
    # finally check the decryption matches the original input
    if plaintext_test == plaintext:
        return 'Test passed'
    return 'Test failed, key = ' + str(key)

def test_vigenere(txt_file_name):
    # first get the plaintext and generate a random key
    plaintext = get_plaintext(txt_file_name)
    key_len = random.randint(1, len(plaintext))
    key = ''
    for i in range(key_len):
        key += ascii_lowercase[random.randint(0, 25)]
    # then encrypt and deccrypt the message
    ciphertext_test =  vigenere_cipher(plaintext, key)
    plaintext_test = decrypt_vigenere_cipher(ciphertext_test, key)
    # finally check the decryption matches the original input
    if plaintext_test == plaintext:
        return 'Test passed'
    return 'Test failed, key = ' + key

# Run both tests on every message file provided for the project.
# A new random shift / key is generated for each file. To test another file just add its name to this list.
message_files = ['message_emma.txt',
                 'message_heart_of_darkness.txt',
                 'message_moby_dick.txt',
                 'message_tomorrows_tangle.txt',
                 'message_treasure_island.txt',
                 'message_wuthering_heights.txt']

for txt_file_name in message_files:
    print(txt_file_name)
    print('  Caesar:   ' + test_caesar(txt_file_name))
    print('  Vigenere: ' + test_vigenere(txt_file_name))
```

```python
# QUESTION 1 - encryption and decryption to and from files
# Run the cell above first, as these functions use the cipher functions defined there.

# a function to write text to a file
def write_text(txt_file_name, text):
    with open(txt_file_name, "w") as file:
        file.write(text)


# Each function below reads the input file, applies the cipher and writes the result to the output file.
# The encryption functions only encrypt the alphabetic content of the input file (see get_plaintext).
# The result is also returned so that it can be used directly.

def encrypt_file_caesar(input_file_name, output_file_name, key):
    ciphertext = caesar_cipher(get_plaintext(input_file_name), key)
    write_text(output_file_name, ciphertext)
    return ciphertext

def decrypt_file_caesar(input_file_name, output_file_name, key):
    plaintext = decrypt_caesar_cipher(get_plaintext(input_file_name), key)
    write_text(output_file_name, plaintext)
    return plaintext

def encrypt_file_vigenere(input_file_name, output_file_name, key):
    ciphertext = vigenere_cipher(get_plaintext(input_file_name), key)
    write_text(output_file_name, ciphertext)
    return ciphertext

def decrypt_file_vigenere(input_file_name, output_file_name, key):
    plaintext = decrypt_vigenere_cipher(get_plaintext(input_file_name), key)
    write_text(output_file_name, plaintext)
    return plaintext


# Now tests which go through files with random keys.
# For a file called name.txt these create the files encrypted_name.txt and decrypted_name.txt.

def test_caesar_files(txt_file_name):
    encrypted_file_name = 'encrypted_' + txt_file_name
    decrypted_file_name = 'decrypted_' + txt_file_name
    key = random.randint(0, 25)
    # encrypt the file and then decrypt the encrypted file
    encrypt_file_caesar(txt_file_name, encrypted_file_name, key)
    decrypt_file_caesar(encrypted_file_name, decrypted_file_name, key)
    # finally check the decrypted file matches the alphabetic content of the original file
    if get_plaintext(decrypted_file_name) == get_plaintext(txt_file_name):
        return 'Test passed'
    return 'Test failed, key = ' + str(key)

def test_vigenere_files(txt_file_name):
    encrypted_file_name = 'encrypted_' + txt_file_name
    decrypted_file_name = 'decrypted_' + txt_file_name
    # generate a random key
    key_len = random.randint(1, len(get_plaintext(txt_file_name)))
    key = ''
    for i in range(key_len):
        key += ascii_lowercase[random.randint(0, 25)]
    # encrypt the file and then decrypt the encrypted file
    encrypt_file_vigenere(txt_file_name, encrypted_file_name, key)
    decrypt_file_vigenere(encrypted_file_name, decrypted_file_name, key)
    # finally check the decrypted file matches the alphabetic content of the original file
    if get_plaintext(decrypted_file_name) == get_plaintext(txt_file_name):
        return 'Test passed'
    return 'Test failed, key = ' + key

# Run both file tests on every message file (message_files is defined in the cell above).
# Note the Vigenere test writes to the same file names as the Caesar test, so the files left over are the Vigenere ones.
for txt_file_name in message_files:
    print(txt_file_name)
    print('  Caesar:   ' + test_caesar_files(txt_file_name))
    print('  Vigenere: ' + test_vigenere_files(txt_file_name))
```

```python
# QUESTION 2
# Run the first cell above first, as this cell uses caesar_cipher, decrypt_caesar_cipher and get_plaintext from Question 1.

# Import books.
book1 = get_plaintext("moby_dick.txt")
book2 = get_plaintext("tale_of_two_cities.txt")
book3 = get_plaintext("treasure_island.txt")
book4 = get_plaintext("wuthering_heights.txt")
book5 = get_plaintext("pride_and_prejudice.txt")
book6 = get_plaintext("pygmailion.txt")    
# Concatenate into one "master text" we can analyse.
master_text = book1 + book2 + book3 + book4 + book5 + book6 
#Declare this globally so it doesn't need to be rerun each time we call the function.

def frequency_analysis(ciphertext):
    # Calculate the frequency of each letter (uppercase or lowercase) in this master text.
    letter_frequencies = []
    for i in range(26):
        letter_sum = master_text.count(ascii_lowercase[i]) + master_text.count(ascii_uppercase[i])
    #Add to a list of tuples, the first element corresponding to the position of a letter in the alphabet and the second to the frequency.
        letter_frequencies.append((i, letter_sum / len(master_text) * 100))

    # Now sort.
    letter_frequencies = sorted(letter_frequencies, key=lambda x: x[1]) 
    # key=lambda x: x[1] allows us to sort the tuples by the second element of the tuple; i.e. the frequency.

    # Count the number of times each letter a-z appears in the ciphertext.
    ciphertext_letter_frequencies = []
    for i in range(len(ascii_lowercase)):
        ciphertext_letter_sum = ciphertext.count(ascii_lowercase[i]) + ciphertext.count(ascii_uppercase[i])
        ciphertext_letter_frequencies.append((i,ciphertext_letter_sum / len(ciphertext) * 100))
    # Now sort
    ciphertext_letter_frequencies = sorted(ciphertext_letter_frequencies, key=lambda x: x[1]) 

    # The key is given by the alphabetic shift between letters of the same frequency.
    # The shift may vary slightly as letter frequencies vary.
    # Hence will look at the shift for the five most common letters and use the one that appears the most. 
    candidates = []
    for i in range(21,26):
        candidates.append((ciphertext_letter_frequencies[i][0] - letter_frequencies[i][0])%26)
    # Then find the most common shift among these five. This is our key.
    key = max(set(candidates), key=candidates.count)
    return decrypt_caesar_cipher(ciphertext, key)

# Now we can encrypt a text to check the function.
plaintext = get_plaintext("message_heart_of_darkness.txt")
plaintext2 = get_plaintext("message_emma.txt")
plaintext3 = get_plaintext("message_moby_dick.txt")
plaintext4 = get_plaintext("message_treasure_island.txt")
plaintext5 = get_plaintext("message_wuthering_heights.txt")
plaintext6 = get_plaintext("message_tomorrows_tangle.txt")
key = random.randint(0, 25)
key2 = random.randint(0,25)
key3 = random.randint(0,25)
key4 = random.randint(0,25)
key5 = random.randint(0,25)
key6 = random.randint(0,25)
ciphertext = caesar_cipher(plaintext, key)
ciphertext2 = caesar_cipher(plaintext2, key2)
ciphertext3 = caesar_cipher(plaintext3, key3)
ciphertext4 = caesar_cipher(plaintext4, key4)
ciphertext5 = caesar_cipher(plaintext5, key5)
ciphertext6 = caesar_cipher(plaintext6, key6)

assert frequency_analysis(ciphertext) == plaintext
assert frequency_analysis(ciphertext2) == plaintext2
assert frequency_analysis(ciphertext3) == plaintext3
assert frequency_analysis(ciphertext4) == plaintext4
assert frequency_analysis(ciphertext5) == plaintext5
assert frequency_analysis(ciphertext6) == plaintext6
```

```python

# Remember to run necessary cells from previous questions before running this cell.
# The functions below are given by the tutorials. See 8.2.1_RSA and 8.2.3 for information.
import sys, os
sys.path.append(os.getcwd())

from random import SystemRandom, randint
## The following functions are given by the tutorials. See 8.2.1_RSA and 8.2.3 for information.
def gcd(a,b):
    """
    Returns the greatest common divisor of integers a and b using Euclid's algorithm.
    The order of a and b does not matter and nor do the signs.
    """
    if not (a % 1 == 0 and b % 1 == 0):
        print( "Need to use integers for gcd.")
        return None
    if b == 0:
        # The gcd is the absolute value of the present value of a.
        return abs(a)                  
    else:
        return gcd(b,a % b)

def random_prime(bit_length):
    '''
    Returns a cryptographically secure random numbber 
    of bit_length many (binary) bits 
    '''
    while True:
        p = SystemRandom().getrandbits(bit_length)  
        if p >= 2**(bit_length-1):
            if is_prime(p):
                return p   
    
def char_to_byte(char):

    byte = bin(ord(char))[2:]
    zeroes = 8 - len(byte)
    byte = '0' * zeroes + byte

    return byte

def integer(char, verbose=False):
    binary = '1'
    binary = binary + char_to_byte(char)
    if verbose:
        return(binary)
    return int(binary, 2)

def RSA_key_generator(bitlength=128):

# The output is ((PublicKey), PrivateExp, Factorisation).

    ## Generate two large primes.
    p = random_prime(bitlength)
    q = random_prime(bitlength)
    N = p*q
    
    ## Next, get phi and use it to get a public and private exponent.
    phi = (p-1)*(q-1)
    ## Choose encryption exponent e randomly between 1 and p and q.
    e = randint(1,min(p,q))
    ## It must be coprime to phi though. We use rejection sampling again:    
    while gcd(e,phi) > 1:
        e = randint(1,min(p,q))
    
    ## Finally, get d using 'pow'
    ## (the extended Euclid's algorithm could do this too).
    d = pow(e,-1,phi)
    
    ## Then return the keys.
    return ((N,e),d%phi,{p:1,q:1})


# My code begins here.


def key_slice(key):

    # First we turn the key into a list of characters and find how many steps we need to do.
    num_key = []
    list_key = [str(i) for i in str(key)]
    for j in list_key:
        num = integer(j)
        num_ten = [str(i) for i in str(num)]
        num_key.extend(num_ten)
        
    step = len(num_key) // 36
    sliced = []
    
    # Null case for if we do not need to slice
    
    if step == 0:
        ''.join(str(l) for l in num_key)
        return [''.join(num_key)]

    # We split every seven digits, until we reach a remainder, then we split the rest. Each time we do this, we join the list back into a string and add it to a list.
    
    else:  
        for n in range(1, step + 2):
            if n<= step + 1:
                x = slice(36*(n-1), 36*n)
                if num_key[x] != []:
                    sliced.append(''.join(str(m) for m in num_key[x]))
        
        return sliced

def sliced_vigenere(txt, key=None):

    # I built the code below using the vigenere test in order to generate my key.
    # If a key is given we use it, otherwise we generate a random one.
    
    plaintext = get_plaintext(txt)
    if key is None:
        key_len = random.randint(1, len(plaintext))
        key = ''
        for i in range(key_len):
            key += ascii_lowercase[random.randint(0, 25)]

    # We apply the cipher and slice it ready for RSA encryption.
        
    cipher_text = vigenere_cipher(plaintext, key)
    vigenere_slice = key_slice(key)
    return vigenere_slice, cipher_text

def rsa_encrypt(txt, pub, key=None):

    # key is the Vigenere key to use. If it is not given a random one is generated.

    # We assign variable names to our vigenere cipher and encrypted text
    
    rsa_vigenere = []
    s, cipher_txt = sliced_vigenere(txt, key)

    # We assign RSA encryption to each slice, and combine the slices back into one key
    
    for t in range(len(s)):
        t = pow(int(s[t]),pub[1],pub[0])
        rsa_vigenere.append(t)
    
    # The encrypted slices of the key are transmitted as a tuple.
    return tuple(rsa_vigenere), cipher_txt

def decrypt(key, cipher_txt, pub, priv):

    # We undo the rsa encryption with the private key.
    
    decrypted = []
    for i in key:
        decrypted.append(str(pow(i, priv, pub[0])))

    # Because we add a 1 to the beginning of the binary representations of each character (see lecture 8.2.3), every letter is represented by a 3 digit number.
    # Therefore we can split our key into groups of three, subtract 256 and apply chr to get our vigenere key.
    # Finally we apply the decryption to decipher the message.

    # One thing to note is that spaces are lost somewhere, maybe in the vigenere cipher?
        
    key = ''
    group = ''.join(decrypted)
    
    for j in range(0, len(group), 3):
        x = slice(j, j+3)
        key = key + chr(int(group[x]) - 256)
        
    raw = decrypt_vigenere_cipher(cipher_txt, key)
    print(raw)
test = 'The Quick Brown Fox Jumps Over The Lazy Dog'  
```
